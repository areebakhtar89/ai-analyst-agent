```python
import os
from huggingface_hub import InferenceClient

HF_TOKEN = os.environ.get("HF_TOKEN")

if not HF_TOKEN:
    raise ValueError("HF_TOKEN not found")

print("Token detected. Testing chat completion...")

try:
    client = InferenceClient(
        model="meta-llama/Llama-3.3-70B-Instruct",
        token=HF_TOKEN,
        provider="auto",
    )

    response = client.chat.completions.create(
        messages=[
            {"role": "user", "content": "What is your name? and how many parameters are you trained upn"}
        ],
        max_tokens=2000,
    )

    print("Model response:")
    print(response.choices[0].message.content)
    print("\nSUCCESS: Token and model access working.")

except Exception as e:
    print("\nERROR:")
    print(str(e))
```

```python
from groq import Groq

client = Groq()
completion = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[
      {
        "role": "user",
        "content": "hey, what is the capital of South Africa?"
      }
    ],
    temperature=1,
    max_completion_tokens=1024,
    top_p=1,
    stream=True,
    stop=None
)

for chunk in completion:
    print(chunk.choices[0].delta.content or "", end="")
```
